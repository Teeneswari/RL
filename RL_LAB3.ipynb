```python
import random
import pandas as pd
```

```python
machines = ['A', 'B', 'C']

M = 3
N_play = 10
epsilon = 0.2

# Initialize N, R and Q
N = {'A': 0, 'B': 0, 'C': 0}
R = {'A': 0, 'B': 0, 'C': 0}
Q = {'A': 0.0, 'B': 0.0, 'C': 0.0}

print("Initial Tables")
print("N =", N)
print("R =", R)
print("Q =", Q)
```

```python
reward_patterns = {
    'A': [3, 4, 2, 5],
    'B': [6, 7, 5, 8],
    'C': [4, 3, 5, 4]
}

# Keeps track of the next reward for each machine
reward_index = {
    'A': 0,
    'B': 0,
    'C': 0
}
```

```python
random.seed(42)
```

```python
results = []
```

```python
for play in range(1, N_play + 1):

    # Generate random number between 0 and 1
    r = random.random()

    # Find highest current Q-value
    max_Q = max(Q.values())

    # Select the first machine having highest Q-value
    best_machine = next(
        machine for machine in machines
        if Q[machine] == max_Q
    )
```

```python
if r < epsilon:

        # During exploration, exclude highest-Q machine
        possible_machines = [
            machine for machine in machines
            if machine != best_machine
        ]

        selected_machine = random.choice(possible_machines)

        decision = "Exploration"

else:
    selected_machine = best_machine
    decision = "Exploitation"

print("Best Machine:", best_machine)
print("Selected Machine:", selected_machine)
print("Decision:", decision)
```

```python
for play in range(1, N_play + 1):

    r = random.random()

    max_Q = max(Q.values())

    best_machine = next(
        machine for machine in machines
        if Q[machine] == max_Q
    )

    if r < epsilon:

        possible_machines = [
            machine for machine in machines
            if machine != best_machine
        ]

        selected_machine = random.choice(possible_machines)

        decision = "Exploration"

    else:

        selected_machine = best_machine

        decision = "Exploitation"
```

```python
for play in range(1, N_play + 1):

    r = random.random()

    max_Q = max(Q.values())

    best_machine = next(
        machine for machine in machines
        if Q[machine] == max_Q
    )

    if r < epsilon:
        possible_machines = [
            machine for machine in machines
            if machine != best_machine
        ]
        selected_machine = random.choice(possible_machines)
        decision = "Exploration"
    else:
        selected_machine = best_machine
        decision = "Exploitation"

    index = reward_index[selected_machine]
    reward = reward_patterns[selected_machine][index]

    reward_index[selected_machine] += 1

    N[selected_machine] += 1
    R[selected_machine] += reward

    Q[selected_machine] = (
        Q[selected_machine]
        + (reward - Q[selected_machine])
        / N[selected_machine]
    )

    results.append({
        'Play': play,
        'Random r': round(r, 4),
        'Decision': decision,
        'Machine': selected_machine,
        'Reward': reward,
        'N(A)': N['A'],
        'N(B)': N['B'],
        'N(C)': N['C'],
        'R(A)': R['A'],
        'R(B)': R['B'],
        'R(C)': R['C'],
        'Q(A)': round(Q['A'], 4),
        'Q(B)': round(Q['B'], 4),
        'Q(C)': round(Q['C'], 4)
    })
```

```python
df = pd.DataFrame(results)

print("\nPlay-by-Play Results:")
display(df)
```

```python
print("\nFinal N Table - Number of Times Each Machine Was Played")

N_table = pd.DataFrame({
    'Machine': machines,
    'N (Participation Count)': [
        N['A'],
        N['B'],
        N['C']
    ]
})

display(N_table)
```

```python
print("\nFinal R Table - Total Rewards")

R_table = pd.DataFrame({
    'Machine': machines,
    'R (Total Reward)': [
        R['A'],
        R['B'],
        R['C']
    ]
})

display(R_table)
```

```python
print("\nFinal Q Table - Estimated Reward")

Q_table = pd.DataFrame({
    'Machine': machines,
    'Q (Estimated Reward)': [
        round(Q['A'], 4),
        round(Q['B'], 4),
        round(Q['C'], 4)
    ]
})

display(Q_table)
```

```python
total_reward = sum(R.values())

print("\nTotal Reward Received =", total_reward)
```

```python
best_machine_final = max(Q, key=Q.get)

print("Machine with Highest Final Q-value =", best_machine_final)
print("Highest Q-value =", round(Q[best_machine_final], 4))
```
