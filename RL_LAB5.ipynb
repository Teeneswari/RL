```python
def random_agent(env_name, episodes=5):

    rewards = []

    print("\n" + "=" * 50)
    print("Environment:", env_name)
    print("=" * 50)

    for episode in range(episodes):

        # 1. Create environment
        env = gym.make(env_name)

        # 2. Reset environment
        state, info = env.reset()

        # 3. Observe initial state
        # 4. Identify available actions
        action_space = env.action_space

        total_reward = 0
        done = False

        # Continue until episode terminates
        while not done:

            # 5. Select a random action
            action = action_space.sample()

            # 6. Execute action
            next_state, reward, terminated, truncated, info = env.step(action)

            # 7. Observe next state and reward
            total_reward += reward

            # Move to next state
            state = next_state

            # 8. Check termination
            done = terminated or truncated

        # 9. Store total reward
        rewards.append(total_reward)

        print("Episode", episode + 1, "-> Total Reward:", total_reward)

        env.close()

    # Calculate average reward
    average_reward = np.mean(rewards)

    print("Rewards:", rewards)
    print("Average Reward:", average_reward)

    return rewards, average_reward
```

```python
cartpole_rewards, cartpole_average = random_agent(
    "CartPole-v1", 5
)
```

```python
mountaincar_rewards, mountaincar_average = random_agent(
    "MountainCar-v0", 5
)
```

```python
frozenlake_rewards, frozenlake_average = random_agent(
    "FrozenLake-v1", 5
)
```

```python
print("\n" + "=" * 50)
print("FINAL RESULTS")
print("=" * 50)

print("\nCartPole")
print("Total Rewards:", cartpole_rewards)
print("Average Reward:", cartpole_average)

print("\nMountainCar")
print("Total Rewards:", mountaincar_rewards)
print("Average Reward:", mountaincar_average)

print("\nFrozenLake")
print("Total Rewards:", frozenlake_rewards)
print("Average Reward:", frozenlake_average)

print("\nExperiment Completed Successfully!")
```
